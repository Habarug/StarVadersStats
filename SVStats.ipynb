```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt
import json
import os
```

```python
runs = {}

for directory in os.listdir("data/history"):
    rundir = os.path.join("data/history/", directory)

    runs[directory] = {}

    with open(os.path.join(rundir, "rundata.json")) as f:
        runs[directory]["rundata"] = json.load(f)

    with open(os.path.join(rundir, "playerdata.json")) as f:
        runs[directory]["playerdata"]= json.load(f)

    with open(os.path.join(rundir, "challengedata.json")) as f:
        runs[directory]["challengedata"] = json.load(f)

pilots = set([run["playerdata"]["PilotName"] for _, run in runs.items()])
_decks = [run["playerdata"]["deckCardDataList"] for _, run in runs.items()]
cards = set([card["Card"] for deck in _decks for card in deck])
_artifactdecks = [run["playerdata"]["artifactList"] for _, run in runs.items()]
artifacts = set([artifact for artlist in _artifactdecks for artifact in artlist])

        
```

```python
fig, ax = plt.subplots()

difficulties = [run["challengedata"]["Difficulty"] for _, run in runs.items()]
victories = [run["rundata"]["isVictory"] for _, run in runs.items()]

sct = ax.scatter(range(len(difficulties)), difficulties, c = victories, marker = "o")
# cbar = plt.colorbar(sct, ax = ax)
# cbar = 

ax.grid()
ax.set_ylabel("Difficulty")
ax.set_xlabel("Runs")
```
